```python
import importlib

transformers = importlib.import_module("transformers")
AutoTokenizer = transformers.AutoTokenizer
GPT2LMHeadModel = transformers.GPT2LMHeadModel
set_seed = transformers.set_seed

# 1. Load pretrained GPT-2 model and tokenizer
model_name = "openai-community/gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# 2. Prompt definition
prompt = "This royal throne of kings, this sceptred isle"

# 3. Tokenize and encode the prompt
inputs = tokenizer(prompt, return_tensors="pt")

# Set seed for reproducibility (optional)
set_seed(42)

# 4 & 5. Generate text with default parameters
print("=== Output Set 1 (temperature=1.0, top_p=0.9) ===")
outputs_1 = model.generate(
    **inputs,
    do_sample=True,
    max_new_tokens=40,
    temperature=1.0,
    top_p=0.9,
    num_return_sequences=5,
    pad_token_id=tokenizer.eos_token_id
)

for i, output in enumerate(outputs_1):
    generated_text = tokenizer.decode(output, skip_special_tokens=True)
    print(f"\nSequence {i+1}:\n{generated_text}")

# 6. Repeat generation with altered temperature (e.g., lower temperature = 0.3)
print("\n" + "="*50)
print("=== Output Set 2 (temperature=0.3, top_p=0.9) ===")
outputs_2 = model.generate(
    **inputs,
    do_sample=True,
    max_new_tokens=40,
    temperature=0.3,
    top_p=0.9,
    num_return_sequences=5,
    pad_token_id=tokenizer.eos_token_id
)

for i, output in enumerate(outputs_2):
    generated_text = tokenizer.decode(output, skip_special_tokens=True)
    print(f"\nSequence {i+1}:\n{generated_text}")
```
